## Exercice 1  Nombre d’enfants à charge par famille

Une enquête réalisée dans un village porte sur le nombre d’enfants à charge par famille. Les résultats sont les suivants :

| Nombre d’enfants | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
|:--|--:|--:|--:|--:|--:|--:|--:|
| Effectif | 18 | 32 | 66 | 41 | 32 | 9 | 2 |

```python
# Données de l’exercice 1
nombre_enfants = [0, 1, 2, 3, 4, 5, 6]
effectifs = [18, 32, 66, 41, 32, 9, 2]
```

**Question 1.** Décrivez la population étudiée, son effectif total, l’unité statistique, le(s) caractère(s) étudié(s), le type de variable et ses valeurs/modalités.

```python
effectif_total = sum(effectifs)
print("Effectif total :", effectif_total)
```

**Question 2.** Calculez les caractéristiques de tendance centrale : moyenne, médiane, mode.

```python
donnees = []

for valeur, effectif in zip(nombre_enfants, effectifs):
    donnees += [valeur] * effectif

moyenne = np.mean(donnees)
mediane = np.median(donnees)

mode = nombre_enfants[effectifs.index(max(effectifs))]

print("Moyenne :", moyenne)
print("Médiane :", mediane)
print("Mode :", mode)
```

**Question 3.** Calculez les caractéristiques de dispersion : étendue, écart interquartile, variance et écart-type.

```python
etendue = max(donnees) - min(donnees)

q1 = donnees_sorted = sorted(donnees)
q1 = donnees_sorted[49]       # 50e valeur
q3 = donnees_sorted[149]      # 150e valeur

ecart_interquartile = q3 - q1

variance = np.var(donnees)
ecart_type = np.std(donnees)

print("Étendue :", etendue)
print("Q1 :", q1)
print("Q3 :", q3)
print("Écart interquartile :", ecart_interquartile)
print("Variance :", variance)
print("Écart-type :", ecart_type)
```

**Question 4.** Tracez l’histogramme, la densité et la boîte à moustaches (boxplot).

```python
plt.figure()
sns.histplot(donnees, discrete=True)
plt.title("Histogramme du nombre d'enfants")
plt.xlabel("Nombre d'enfants")
plt.ylabel("Effectif")
plt.show()

plt.figure()
sns.histplot(donnees, discrete=True, stat="density", kde=True)
plt.title("Densité du nombre d'enfants")
plt.xlabel("Nombre d'enfants")
plt.show()

plt.figure()
sns.boxplot(x=donnees)
plt.title("Boîte à moustaches")
plt.xlabel("Nombre d'enfants")
plt.show()
```

**Question 4.** Tracez l’histogramme, la densité et la boîte à moustaches (boxplot).

## Exercice 5  Diamètre de tubes plastiques

Un échantillon de 100 tubes plastiques a été prélevé. Le diamètre, mesuré en décimètres, est regroupé par intervalles :

| Diamètre (dm) | [1.9, 2.0[ | [2.0, 2.1[ | [2.1, 2.2[ | [2.2, 2.3[ | [2.3, 2.4[ | [2.4, 2.5[ | [2.5, 2.6[ |
|:--|--:|--:|--:|--:|--:|--:|--:|
| Effectif | 3 | 9 | 18 | 29 | 25 | 10 | 6 |

```python
# Données de l’exercice 5
intervalles_diametre = [(1.9, 2.0), (2.0, 2.1), (2.1, 2.2), (2.2, 2.3), (2.3, 2.4), (2.4, 2.5), (2.5, 2.6)]
effectifs = [3, 9, 18, 29, 25, 10, 6]
```

**Question 1.** Décrivez la population étudiée, son effectif total, l’unité statistique, le(s) caractère(s) étudié(s), le type de variable et ses valeurs/modalités.

```python
import numpy as np

intervalles_diametre = [
    (1.9, 2.0),
    (2.0, 2.1),
    (2.1, 2.2),
    (2.2, 2.3),
    (2.3, 2.4),
    (2.4, 2.5),
    (2.5, 2.6)
]

effectifs = [3, 9, 18, 29, 25, 10, 6]

centres = []

for intervalle in intervalles_diametre:
    centre = (intervalle[0] + intervalle[1]) / 2
    centres.append(centre)

print("Centres :", centres)


donnees = np.repeat(centres, effectifs)

print("Effectif total :", len(donnees))
```

**Question 2.** Calculez les caractéristiques de tendance centrale : moyenne, médiane, mode.

```python

moyenne = np.mean(donnees)
mediane = np.median(donnees)

mode = centres[3]

print("Moyenne :", moyenne)
print("Médiane :", mediane)
print("Mode :", mode)
```

**Question 3.** Calculez les caractéristiques de dispersion : étendue, écart interquartile, variance et écart-type.

```python
print("Étendue :", np.max(donnees) - np.min(donnees))
print("Q1 :", np.percentile(donnees, 25))
print("Q3 :", np.percentile(donnees, 75))
print("Écart interquartile :",np.percentile(donnees, 75) - np.percentile(donnees, 25))
print("Variance :", np.var(donnees))
print("Écart-type :", np.std(donnees))
```

**Question 4.** Tracez l’histogramme, la densité et la boîte à moustaches (boxplot).

```python
bornes = [intervalles_diametre[0][0]]

for intervalle in intervalles_diametre:
    bornes.append(intervalle[1])

print(bornes)



plt.hist(donnees, bins=bornes, edgecolor="black")
plt.xlabel("Diamètre (dm)")
plt.ylabel("Effectif")
plt.title("Histogramme des diamètres")
plt.show()

# Densité
plt.hist(donnees, bins=bornes, density=True, edgecolor="black")
plt.xlabel("Diamètre (dm)")
plt.ylabel("Densité")
plt.title("Densité des diamètres")
plt.show()

# Boîte à moustaches
plt.figure()
sns.boxplot(x=donnees)
plt.xlabel("Diamètre (dm)")
plt.title("Boîte à moustaches")
plt.show()
```
